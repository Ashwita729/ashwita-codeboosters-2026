```python
#Import all the libaries
import pandas as pd
import numpy as np
#New-library - sentence-transformers
#sentence-transformers : the class we used to load the embedding model
from sentence_transformers import SentenceTransformer

#new library
#chromadb : the vector database package
import chromadb

print("All libraries imported successfully")
print(f'Chroma version : {chromadb.__version__}')

```

```python
# Keyword Search vs Semantic Search

documents = [
    "ETL is used to clean and tranform the data",   #doc_1
    "A vehicle is a mode of transportation",        #doc_2
    "Cars and trucks are popular automobiles",      #doc_3
    "SQL is used to query databases",               #doc_4
    "Machine Learning trains models on data",       #doc_5
]

#---keyword seearch---
#keyword search : checks if te exact query word appears in the document


query_keyword = "vehicle"  #the word we are searching for

print('='*60)
print(f'KEYWORD SEARCH for : {query_keyword}')
print('='*60)

for i,doc in enumerate(documents):
  #.lower() makes comparison case-sensitive
  #'in' checks
  if query_keyword.lower() in doc.lower():
    print(f'FOUND  [doc_{i}] : {doc}')
  else:
    print(f'MISSED [doc_{i}] : {doc}')



print()
print("PROBLEM : doc_2 talks about 'Cars and trucks' - which are vehicles!")
print("But keyword search missed it because it only checks for 'vehicle'")
```

```python
#Load the embedding model
#all- means trained on many different types of text
#miniLM : a smaller version

print("Loading embedding model...(may take 1-2 minutes onf in first run)")
model = SentenceTransformer('all-MiniLM-L6-v2')

print("Embedding model loaded successfully")
print(f"Model produces vectors of size : {model.get_sentence_embedding_dimension()}")
```

```python
#Generate First embedding

#Define a single sentence to embed
sentence= "ETL is used to clean and transform data"

#model.encode() converts text to a vector(list of numbers)
#Input : list of string or string
#Output: NumPy array
embedding = model.encode(sentence)

print(f"Input sentence : '{sentence}'")
print()
print(f"Embedding Type : {type(embedding)}")

print(f"Embedding Shape : {embedding.shape}")
#shape (384,) means: 1 sentence x 384 numbers

print(f"First 15 numbers : {embedding[:15].round(4)}")


print(f"Min value: {embedding.min():.4f}")
print(f"Max value: {embedding.max():.4f}")
#the value ranges roughly from -1 to 1
```

```python
sentences = [
    "ETL is used to clean and tranform the data",   #sentence 1
    "A vehicle is a mode of transportation",        #sentence 2
    "The sky is blue and clouds are white"          #sentence 3
]
embedding = model.encode(sentences)

print(f"Number of sentences: {len(sentences)}")
print(f"Embedding shape : {embedding.shape}")

print()
print("Each row is one sentence's embeddings : ")
for i, sent in enumerate(sentences):
  print(f"Sentences {i} : shape(embeddings[i].shape), first 5 values : {embedding[i][:5]}")
```

```python
from sklearn.metrics.pairwise import cosine_similarity

your_sentences=[
    "Biology is studied to be a doctor",
    "Shreya likes biology",
    "I enjoy eating pizza for dinner",
]

your_embeddings = model.encode(your_sentences)

sim_your_01 = cosine_similarity(your_embeddings[0].reshape(1, -1),your_embeddings[1].reshape(1, -1))[0][0]
sim_your_02 = cosine_similarity(your_embeddings[0].reshape(1, -1),your_embeddings[2].reshape(1, -1))[0][0]

print("Your Experiment results : ")
print('='*60)
print(f"Sentence A : {your_sentences[0]}")
print(f"Sentence B : {your_sentences[1]}")
print(f"Sentence C : {your_sentences[2]}")
print(f"Similarity (A vs B) : {sim_your_01:.4f}")
print(f"Similarity (A vs C) : {sim_your_02:.4f}")
print()
```

```python

chroma_client = chromadb.Client()
#chromadb.Client() creates and IN-MEMORY database
#in-memory: data exists only while the notebook is running
#When you restart the kernel, the data is gone an dyou must re-add it

collection = chroma_client.get_or_create_collection('demo_notes')
#demo_note is the name we give to the collection

print("ChromaDB Client created (in-memory mode)")
print(f"Collection name : demo notes")
print(f"Documents in collection : {collection.count()}")
#count returns how many documents are currently in this collection
#should be 0 because we just created (or if it existed, its previous count)
```

```python

sample_docs = [
    "ETL is used to clean and transform data",
    "A vehicle is a mode of transportation",
    "Cars and trucks are popular automobiles",
    "SQL is used to query databases",
    "Machine learning trains models on data"
]

sample_ids = ['doc001', 'doc002', 'doc003', 'doc004', 'doc005']

sample_metadata = [
    {"subject": "Data Engineering", "topic": "ETL"},
    {"subject": "Data Engineering", "topic": "SQL"},
    {"subject": "Machine Learning", "topic": "ML Basic"},
    {"subject": "Python", "topic": "Pandas"},
    {"subject": "Machine Learning", "topic": "Neural Networks"}
]

sample_embeddings = model.encode(sample_docs).tolist()

collection.add(
    documents=sample_docs,
    embeddings=sample_embeddings,
    ids=sample_ids,
    metadatas=sample_metadata
)

print("Documents added successfully")
print(f"Number of documents in collection: {collection.count()}")
```

```python


query = "How do i clean and prepare data?"

results = collection.query(
    query_texts=[query], # Corrected: pass a list containing the query string
    n_results=3         #returns top 3 similar docs
)

print("RESULT KEYS ARE AVAILABLE")
print(list(results.keys())) # Corrected: use 'results' variable
```

```python
#Display the results in a Readable format

print(f"Query : '{query}'")
print('='*60)
print()

matched_docs = results['documents'][0]
matched_ids = results['ids'][0]
matched_distances = results['distances'][0]
matched_metadata = results['metadatas'][0]

for rank,(doc,doc_id,distance,metadata) in enumerate(zip(matched_docs,matched_ids,matched_distances,matched_metadata)):
  print(f"Rank {rank} | ID : {doc_id} | Distance : {distance:.4f}")
  #Distance lower - More similar
  print(f" Subject : {metadata['subject']} | Topic : {metadata['topic']}")
  print(f"Document : {doc}")
  print()

print("NOTICE : The results are about ETL and Pandas - excactly what ' clean and prepare data ' ")
print("Semantic search found them even though the words are different")
```

```python
filtered_query = "How do computers learn from samples?"
filtered_results = collection.query(
    query_texts=[filtered_query],
    n_results=3,
    where={"subject" : "Machine Learning"}
)

print(f"Filtered Query : '{filtered_query}'")
print("Filter : Only Machine learning")
print('='*60)

for rank,(doc,dist,metadata) in enumerate(zip(
  filtered_results['documents'][0],
  filtered_results['distances'][0],
  filtered_results['metadatas'][0]
),start=1):
   print(f"Rank {rank} | Distance : {dist:.4f} | Subject : {metadata['subject']} ")
   print(f"Document : {doc}")
   print()

print("Only ML Documents appear, event though ETL and Pandas might be 'related')")
```

```python
print("DISTANCE TO SIMILARITY CONVERSION")
print('='*60)
print(f'{'Distance':<15} {'Similarity':<15} {'Interpretation':<15}')
print('='*60)

distances = [0.05,0.20,0.40,0.65,0.90]
interpretation = ["Wear identical", "Very similar","Related","Not related","Somewhat related"]

for d, interp in zip(distances, interpretation):
  similarity = 1 - d
  print(f"{d:<15.2f} {similarity:<15.2f} {interp:<20}")
```
